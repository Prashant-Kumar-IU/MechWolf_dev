# MechWolf Network Apparatus Designer Demo

This notebook demonstrates the new network-based apparatus designer that allows you to visually create MechWolf apparatus configurations by dragging and dropping components, creating connections, and automatically generating the corresponding `A.add()` code.

## Features
- **Visual Component Library**: Drag components from organized library
- **Network Canvas**: Visual representation of your apparatus
- **Properties Panel**: Configure component settings and connections
- **Automatic Code Generation**: Produces clean MechWolf `A.add()` calls
- **Network Validation**: Ensures apparatus is properly connected
- **Save/Load Configurations**: Persist your designs for reuse

## Setup and Imports

````python
# Import required modules
import sys
import os

# Add MechWolf to path if needed
sys.path.append('/home/kumar/MechWolf')

import mechwolf as mw
from mechwolf.DataEntry.Phase2_ApparatusBuilder.network_apparatus_designer import (
    NetworkApparatusDesigner, 
    create_network_apparatus_designer
)

print("✅ MechWolf Network Apparatus Designer loaded successfully!")
````

## Launch the Network Designer

Run the cell below to launch the interactive network apparatus designer. The interface consists of four main panels:

1. **Left Panel**: Component library with passive and active components
2. **Center Panel**: Network canvas showing your apparatus design
3. **Right Panel**: Properties configuration for selected components
4. **Bottom Panel**: Generated MechWolf code preview

````python
# Create and display the network designer
designer = create_network_apparatus_designer()
designer.display()
````

## How to Use the Designer

### Step 1: Add Components
- Click buttons in the **Component Library** (left panel) to add components to your apparatus
- Components are automatically named (e.g., `vessel_1`, `pump_1`, etc.)
- Both passive components (vessels, mixers) and active components (pumps, valves) are available

### Step 2: Create Connections
- Use the connection controls in the **Network Canvas** (center panel)
- Select a "From" component and "To" component from the dropdowns
- Click "Add Connection" to create a tube connection between them

### Step 3: Configure Properties
- Component-specific properties can be configured in the **Properties Panel** (right panel)
- Set serial ports, tube dimensions, and other component parameters
- Changes are applied when you click "Apply Changes"

### Step 4: Generate Code
- The **Code Preview Panel** (bottom) automatically updates with generated MechWolf code
- Click "⚡ Generate Code" to refresh the code
- Copy the generated code to use in your MechWolf scripts

## Example: Recreating the Birch Reduction Apparatus

Let's recreate the Birch Reduction apparatus from the original manual code using the network designer:

### Original Manual Code:
```python
A = mw.Apparatus("Birch Reduction")
A.add(THF, reagent_valve, fat_tube("1 ft"))
A.add(SM, reagent_valve, fat_tube("1 ft"))
A.add(reagent_valve, reagent_pump, valve_tube)
A.add(reagent_pump, T_mixer, thinner_tube("1 ft"))
A.add(EDAinTHF, Li_activator_pump, fat_tube("1 ft"))
A.add(Li_activator_pump, column, thinner_tube("1 ft"))
A.add(column, T_mixer, thinner_tube("0.16 ft"))
A.add(T_mixer, Product, thinner_tube("15 ft"))
```

### Using the Network Designer:
1. **Add Components**:
   - Click "📦 Vessel" 3 times to create THF, SM, EDAinTHF vessels
   - Click "📦 Vessel" 1 more time for Product vessel
   - Click "🔀 ViciValve" to add reagent_valve
   - Click "🔧 VarianPump" twice to add reagent_pump and Li_activator_pump
   - Click "🔀 TMixer" to add T_mixer
   - Click "📦 Vessel" for column (or add as passive component)

2. **Create Connections**:
   - THF → reagent_valve (fat_tube, 1 ft)
   - SM → reagent_valve (fat_tube, 1 ft)
   - reagent_valve → reagent_pump (valve_tube)
   - reagent_pump → T_mixer (thinner_tube, 1 ft)
   - EDAinTHF → Li_activator_pump (fat_tube, 1 ft)
   - Li_activator_pump → column (thinner_tube, 1 ft)
   - column → T_mixer (thinner_tube, 0.16 ft)
   - T_mixer → Product (thinner_tube, 15 ft)

3. **Configure Properties**:
   - Set serial ports for pumps and valve
   - Configure valve mapping: THF→4, SM→5
   - Set pump flow rates and syringe specifications

4. **Generate and Copy Code**:
   - The designer will generate equivalent MechWolf code
   - Copy the code for use in your protocols

## Advanced Features

### Network Validation
The designer includes built-in validation to check:
- All components are properly connected
- No orphaned vessels or components
- Network connectivity and flow paths
- Potential issues with apparatus design

Click the "✅ Validate" button in the network canvas to check your design.

### Component Library
The designer includes all major MechWolf components:

**Passive Components:**
- Vessels (reagent containers)
- T-Mixers, Y-Mixers, Cross-Mixers
- Custom passive components

**Active Components:**
- FreeStep Pumps (3D syringe pumps)
- Harvard Pumps (syringe pumps)
- Varian Pumps (HPLC pumps)
- VICI Valves (multi-port valves)

**Tube Types:**
- fat_tube: 1/16" ID, 1/8" OD, PFA
- thin_tube: 0.030" ID, 1/16" OD, PFA
- thinner_tube: 0.020" ID, 1/16" OD, PFA
- custom_tube: Configurable dimensions
