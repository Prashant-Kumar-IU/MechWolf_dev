## Example: Building a Complex Multi-Stage Synthesis Apparatus

Let's create a complex apparatus for multi-stage organic synthesis:

### Target Apparatus Configuration:
```
Reagent_A ──┐
             ├── Valve_1 ── Pump_1 ──┐
Reagent_B ──┘                        ├── T_Mixer_1 ── Reactor_1 ──┐
                                      │                             │
Catalyst ──── Pump_2 ─────────────────┘                             ├── T_Mixer_2 ── Product
                                                                     │
Quench ────── Pump_3 ────────────────────────────────────────────────┘
```

### Step-by-Step Construction:

1. **Add Components** (click these buttons in the component library):
   - 📦 Vessel × 4 (Reagent_A, Reagent_B, Catalyst, Quench)
   - 📦 Vessel × 2 (Reactor_1, Product)
   - 🔀 ViciValve × 1 (Valve_1)
   - 🔧 VarianPump × 3 (Pump_1, Pump_2, Pump_3)
   - 🔀 TMixer × 2 (T_Mixer_1, T_Mixer_2)

2. **Create Connections** (use connection controls):
   - Reagent_A → Valve_1 (fat_tube, 1 ft)
   - Reagent_B → Valve_1 (fat_tube, 1 ft)
   - Valve_1 → Pump_1 (thin_tube, 6 in)
   - Pump_1 → T_Mixer_1 (thin_tube, 2 ft)
   - Catalyst → Pump_2 (fat_tube, 1 ft)
   - Pump_2 → T_Mixer_1 (thin_tube, 1 ft)
   - T_Mixer_1 → Reactor_1 (thinner_tube, 10 ft)
   - Reactor_1 → T_Mixer_2 (thin_tube, 1 ft)
   - Quench → Pump_3 (fat_tube, 1 ft)
   - Pump_3 → T_Mixer_2 (thin_tube, 6 in)
   - T_Mixer_2 → Product (thin_tube, 2 ft)

3. **Configure Properties** (select each component and set):
   - **Valve_1**: Serial port, mapping {Reagent_A: 1, Reagent_B: 2}
   - **Pump_1**: Serial port, flow rate 2 mL/min
   - **Pump_2**: Serial port, flow rate 0.5 mL/min
   - **Pump_3**: Serial port, flow rate 1 mL/min

4. **Validate and Generate**:
   - Click "✅ Validate" to check the network
   - Click "📊 Stats" to see component breakdown
   - Click "⚡ Generate" to create the apparatus code

### 📊 Network Analysis

The enhanced designer provides detailed network analysis:

````python
# Access network analysis features
if hasattr(enhanced_designer, 'network') and enhanced_designer.network.nodes:
    network = enhanced_designer.network
    
    print("📊 Current Network Analysis:")
    print(f"  • Total nodes: {len(network.nodes)}")
    print(f"  • Total connections: {len(network.connections)}")
    
    # Component type breakdown
    from collections import defaultdict
    comp_types = defaultdict(int)
    for node in network.nodes.values():
        comp_types[node.component_type] += 1
    
    print("\n🔧 Component Breakdown:")
    for comp_type, count in comp_types.items():
        print(f"  • {comp_type}: {count}")
    
    # Validation status
    errors = network.validate_network()
    print(f"\n✅ Validation Status: {'✅ PASSED' if not errors else '❌ ISSUES FOUND'}")
    if errors:
        for error in errors:
            print(f"  • {error}")
else:
    print("🏗️ Build an apparatus first to see network analysis")
````
